## Load Important Functoin

```python
class Data:
    def __init__(self, array_list):
        self.array_list = array_list

    def save(self, filename):
        with open(filename, 'wb') as f:
            pickle.dump(self.array_list, f)

    @classmethod
    def load(cls, filename):
        with open(filename, 'rb') as f:
            array_list = pickle.load(f)
        return cls(array_list)
    
```

### omg8rep

```python
from tqdm import tqdm

methods_path = f"{metadynamics_path}/autocoorel_data_omg8rep"

meta_omg8rep_t2_means = []
meta_omg8rep_t2_stds =[]

for i in tqdm(range(8)):
    omega_loaded_single = Data.load(f"{methods_path}/stationary_distributions{i}.pkl").array_list
    t2_means_single = omega_loaded_single[4]
    t2_uppers_single = omega_loaded_single[5]
    t2_lowers_single = omega_loaded_single[6]
    t2_stds_single = omega_loaded_single[7]

    meta_omg8rep_t2_means.append(t2_means_single)
    meta_omg8rep_t2_stds.append(t2_stds_single)

meta_omg8rep_lagtimes_ns = omega_loaded_single[8]

#meta_omg8rep_pi_all_means = meta_omg8rep_pi_means
meta_omg8rep_t2_means = np.array(meta_omg8rep_t2_means)
meta_omg8rep_t2_stds = np.array(meta_omg8rep_t2_stds)


# Mean across all replicas
meta_omg8rep_t2_all_means = np.nanmean(meta_omg8rep_t2_means, axis=0)

# Law of total variance
within_var = np.nanmean(meta_omg8rep_t2_stds**2, axis=0)  # average within-replica variance
between_var = np.nanvar(meta_omg8rep_t2_means, axis=0, ddof=1)  # var
meta_omg8rep_t2_all_stds = np.sqrt(within_var + between_var)

meta_omg8rep_lagtimes_ns[-1]
```

```python
dt_in_ns = 0.0004

plt.figure()
plt.plot(meta_omg8rep_lagtimes_ns, meta_omg8rep_t2_all_means*dt_in_ns, 'ko-', alpha=1)
plt.fill_between(meta_omg8rep_lagtimes_ns, 
                (meta_omg8rep_t2_all_means + meta_omg8rep_t2_all_stds) *dt_in_ns, 
                 (meta_omg8rep_t2_all_means - meta_omg8rep_t2_all_stds) *dt_in_ns, alpha=0.3)
plt.xlabel('lagtime (ns)')
plt.ylabel('implied timescale (ns)')

plt.yscale('log')
plt.ylim(0.001, 100)
plt.title(f'Implied timescales for Plain MD')
plt.show()
```

## Combined Plot

```python
import matplotlib.pyplot as plt

# ==========================================
# 1. GLOBAL PLOT SETTINGS & COLORS
# ==========================================
MARKER_STYLE = '.'       # 'o' for dots, 's' for squares, None for no markers
LINE_STYLE   = '-'       # '-' for solid, '--' for dashed
LINE_WIDTH   = 2.5       # Thickness of the main lines
LINE_ALPHA   = 1.0       # Transparency of the main lines (0 to 1)
FILL_ALPHA   = 0.3      # Transparency of the shaded error regions (0 to 1)

# Ensure these keys EXACTLY match the "label" values in the datasets below
method_colors = {
    "Plain MD":       "#1f77b4",  # blue
    "EE":             "#2ca02c",  # green
    "HREX":           "#ff7f0e",  # orange
    "REDs":           "#d62728",  # red 
    "Meta OMG":       "#9467bd",  # purple
    "Meta OPH":       "#e377c2",  # pink (added)
    "Meta OPP":       "#7f7f7f",  # gray (added)
    "Meta OPPC":      "#8c564b",  # brown
    "Meta OMG 8rep":  "#bcbd22",  # olive (added)
    "Meta OPPC 8rep": "#17becf"   # cyan (added)
}

# ==========================================
# 2. DATA CONFIGURATION
# ==========================================
datasets = [
    {
        "label": "Plain MD", "dt": 0.01,
        "x": plain_md_lagtimes_ns, "y": plain_md_t2_all_means,
        "upper": plain_md_t2_all_uppers, "lower": plain_md_t2_all_lowers
    },
    {
        "label": "EE", "dt": 0.01,
        "x": ee_lagtimes_ns, "y": ee_t2_all_means,               
        "upper": ee_t2_all_uppers, "lower": ee_t2_all_lowers     
    },
    {
        "label": "HREX", "dt": 0.01,
        "x": hrex_lagtimes_ns, "y": hrex_t2_all_means,
        "upper": hrex_t2_all_uppers, "lower": hrex_t2_all_lowers
    },
    {
        "label": "REDs", "dt": 0.004,
        "x": reds_lagtimes_ns, "y": reds_t2_all_means,
        "upper": reds_t2_all_uppers, "lower": reds_t2_all_lowers
    },
    {
        "label": "Meta OMG", "dt": 0.0004,
        "x": meta_omg_lagtimes_ns, "y": meta_omg_t2_all_means,
        "upper": meta_omg_t2_all_uppers, "lower": meta_omg_t2_all_lowers
    },
    {
        "label": "Meta OPH", "dt": 0.0004,
        "x": meta_oph_lagtimes_ns, "y": meta_oph_t2_all_means,
        "upper": meta_oph_t2_all_uppers, "lower": meta_oph_t2_all_lowers
    },
    {
        "label": "Meta OPP", "dt": 0.0004,
        "x": meta_opp_lagtimes_ns, "y": meta_opp_t2_all_means,
        "upper": meta_opp_t2_all_uppers, "lower": meta_opp_t2_all_lowers
    },
    {
        "label": "Meta OPPC", "dt": 0.0004,
        "x": meta_oppc_lagtimes_ns, "y": meta_oppc_t2_all_means,
        "upper": meta_oppc_t2_all_uppers, "lower": meta_oppc_t2_all_lowers
    },
    {
        "label": "Meta OMG 8rep", "dt": 0.0004,
        "x": meta_omg8rep_lagtimes_ns, "y": meta_omg8rep_t2_all_means,
        "upper": meta_omg8rep_t2_all_means + meta_omg8rep_t2_all_stds, 
        "lower": meta_omg8rep_t2_all_means - meta_omg8rep_t2_all_stds
    },
    {
        "label": "Meta OPPC 8rep", "dt": 0.0004,
        "x": meta_oppc8rep_lagtimes_ns, "y": meta_oppc8rep_t2_all_means,
        "upper": meta_oppc8rep_t2_all_means + meta_oppc8rep_t2_all_stds, 
        "lower": meta_oppc8rep_t2_all_means - meta_oppc8rep_t2_all_stds
    }
]

# ==========================================
# 3. MASTER PLOTTING LOOP
# ==========================================
fig, ax = plt.subplots(figsize=(12, 6))

for data in datasets:
    dt = data["dt"]
    x = data["x"]
    y = data["y"] * dt
    y_upper = data["upper"] * dt
    y_lower = data["lower"] * dt
    
    # Map the color directly from our dictionary
    color = method_colors[data["label"]]
    
    # Plot the line and markers
    ax.plot(x, y, label=data["label"], color=color, 
            marker=MARKER_STYLE, linestyle=LINE_STYLE, 
            alpha=LINE_ALPHA, linewidth=LINE_WIDTH)
             
    # Plot the shaded confidence intervals
    ax.fill_between(x, y_upper, y_lower, color=color, alpha=FILL_ALPHA)

# ==========================================
# 4. FORMATTING & LEGEND
# ==========================================
ax.set_xlabel('lagtime (ns)', fontsize=18)
ax.set_ylabel('implied timescale', fontsize=18)
ax.set_yscale('log')
ax.set_xscale('log')
ax.set_ylim(0.01, 5000)
ax.set_yticks([ 0.1, 10, 1000, 500])
ax.set_title('Implied Timescales', fontsize=20)
# Change tick font size specifically for this plot
ax.tick_params(axis='both', which='major', labelsize=16)

# Extract handles and labels for the custom legend
handles, labels = ax.get_legend_handles_labels()
# Extract handles and labels for the custom legend
handles, labels = ax.get_legend_handles_labels()

# Apply your custom legend at the Figure level
leg = fig.legend(
    handles, 
    labels, 
    loc='upper center',         
    bbox_to_anchor=(0.5, 1),  # Positioned at the very top
    ncol=5,                      # 5 columns creates a neat 2-row legend
    frameon=False, 
    handlelength=1.5,
    handletextpad=0.5,
    columnspacing=1.0,
    fontsize=16
)

# Push the plot down slightly so the title and top-legend don't overlap
plt.subplots_adjust(top=0.82)

plt.show()
```
